```python
#W7A16
lst = list(map(int, input().strip().split()))
dic = {}
count = 0
for i in range(len(lst)):
    if lst[i] == 0:
        count += 1
    if lst[i] not in dic:
        dic[lst[i]] = 1
    else:
        dic[lst[i]] += 1
    if dic[lst[i]] > lst[i]:
        count += 1

if count == 0:
    print("YES")
else:
    print("NO", count)
```

```python
#W7A18
lst = list(map(int, input().split()))
chan = []
le = []
sap_xep = []
for i in lst:
    if i % 2 == 0:
        chan.append(i)
    else:
        le.append(i)

def sap_xep_ds(lst):
    for _ in range(len(lst)):
        flag = False
        for j in range(0, len(lst) - 1):
            if lst[j] > lst[j + 1]:
                lst[j], lst[j + 1] = lst[j + 1], lst[j]
                flag = True
        if flag is False:
            return lst
            break

chan = sap_xep_ds(chan)
le = sap_xep_ds(le)
le.reverse()
print(chan + le)
```

```python
#W7A27
mo = ["[", "(", "{"]
dong = ["]", ")", "}"]
lst = []
day = input()
flag = True
for i in day:
    if i in mo:
        lst.append(i)
    else:
        if i in dong:
            if len(lst) == 0:
                flag = False
                break
            else:
                dau = lst[-1]
                n = mo.index(dau)
                if dong[n] != i:
                    flag = False
                    break
                lst.pop()

if flag == False or len(lst) > 0:
    print("NO")
else:
    print("YES")
```

```python
#W7A28
N = int(input())
from math import isqrt
def snt_check(x):
    for i in range(2, isqrt(x) + 1):
        if x <= 3:
            return False
        if x % i == 0:
            return False
    if x >= 1:
        return True

lst = []
ds = []
for j in range(N+1):
    if snt_check(j) == False:
        lst.append(j)
    elif j == N:
        ds.append(lst)
    elif len(lst) > 0:
        ds.append(lst)
        lst = []
day = []
for k in ds:
    if len(k) > len(day):
        day = k

print(*day, sep = ', ')
```

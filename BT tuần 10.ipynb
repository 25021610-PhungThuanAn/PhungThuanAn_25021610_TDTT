```python
#W9A1
list1 = list(map(int, input().split()))
list2 = list(map(int, input().split()))
kq = []
i, j = 0, 0
while i < len(list1) and j < len(list2):
    if list1[i] > list2[j]:
        kq.append(list2[j])
        j += 1
    else:
        kq.append(list1[i])
        i += 1
kq.extend(list2[j:])
kq.extend(list1[i:])

print(kq)
```

```python
#W9A7
def canPlaceFlowers(flowerbed, k):
    if len(flowerbed) == 1 and flowerbed[0] == 0:
        k -= 1
    if len(flowerbed) > 1:
        if flowerbed[0] == 0 and flowerbed[1] == 0:
            k -= 1
            flowerbed[0] = 1
        if flowerbed[-1] == flowerbed[-2] == 0:
            k -= 1
            flowerbed[-1] = 1
        for i in range(1, len(flowerbed) - 1):
            if flowerbed[i] == 0 and flowerbed[i - 1] == flowerbed[i + 1] == 0:
                flowerbed[i] = 1
                k -= 1
    if k <= 0:
            return True
    return False
flowerbed = list(map(int, input().split()))
k = int(input())
print(canPlaceFlowers(flowerbed, k))
```

```python
#W9A8
moves = list(input())
def is_path_crossing(moves):
    vi_tri = {(0,0)}
    x = 0
    y = 0
    for i in moves:
        if i == "R":
            x += 1
        elif i == "L":
            x -= 1
        elif i == "U":
            y += 1
        elif i == "D":
            y -= 1
        if (x,y) in vi_tri:
            return True
        vi_tri.add((x,y))
    return False
print(is_path_crossing(moves))
```

```python
#W9A9
def Median(nums):
    a, b, c, d, e = nums[0], nums[1], nums[2], nums[3], nums[4]
    if a > b: a, b = b, a
    if c > d: c, d = d, c
    if a > c: 
        a, b, c, d = c, d, a, b
    if e < b: 
        b, e = e, b 
    if b < c:
        if c < e: return c
        return e
    else:
        if c < e:
             if b < d: return b
             return d
        else:
             return c

nums = list(map(int, input().split()))
print(Median(nums))
```

```python
#W9A10
def findRadius(houses, heaters):
    max_dis = []
    for i in houses:
        dis = []
        for j in heaters:
            dis.append(abs(i - j))
        max_dis.append(min(dis))
    return max(max_dis)

houses = list(map(int, input().split()))
heaters = list(map(int, input().split()))
print(findRadius(houses, heaters))
```
